```python
flights_final["flight_key"] = (
    flights_final.index + 1
).astype(str).map(lambda x: "FL" + x.zfill(4))

print(flights_final[["flight_id", "flight_key"]].head())
```

```python
flight_id_reference = (
    flights_final
    .groupby("flight_id")
    .agg(
        flight_record_count=("flight_key", "count")
    )
    .reset_index()
)

flight_id_reference["flight_id_conflict"] = (
    flight_id_reference["flight_record_count"] > 1
)

print(flight_id_reference.head())
print(
    "Conflicting flight IDs:",
    flight_id_reference["flight_id_conflict"].sum()
)
```
